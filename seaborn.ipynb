## Introdução ao Seaborn

<details>
  <summary>

- Biblioteca Python para visualização de dados

<details>
  <summary>
    <font size='3', color='white'>
      <b>Scatter Plot</b>
    </font>
  </summary>
 

- Cada ponto uma observação
- Mostra relação entre duas variáveis
</details>

````python
import seaborn as sns
import matplotlib.pyplot as plt

height = [62, 64, 69, 75, 66, 68, 65, 71, 76, 73]
weight = [120, 136, 148, 175, 137, 165, 154, 172, 200, 187]

sns.scatterplot(x=height, y=weight)
plt.show()
````

<details>

    </font>
  </summary>
 

- é uma função utilizada para visualizar a contagem de observações em cada categoria de uma variável categórica.
- *hue* = terceira variável (cores)

</details>

````python
import pandas as pd

df = pd.DataFrame({
    "genero":  ["Feminino", "Masculino", "Feminino", "Masculino", "Feminino", "Masculino"],
    "idade":   [23, 28, 35, 40, 47, 52],
    "salario": [2800, 4200, 5300, 7800, 6100, 9200]
})
df
````

````python
gender = ["Female","Female","Female","Female","Male","Male","Male","Male","Male","Male"]

sns.countplot(x=gender)
plt.show()
````

````python
sns.scatterplot(x="idade", y="salario", hue="genero", data=df)
plt.title("Salário por Idade e Sexo")
plt.xlabel("Idade")
plt.ylabel("Salário (R$)")
plt.tight_layout()
plt.show()
````

## Relational Plots

<details>
  <summary>
    <font size='3', color='white'>
      <b>Relacionamento entre variáveis</b>
    </font>
  </summary>

- Exemplos de relacionamento entre variaveis:
    * altura vs peso
    * faltas vs nota
- relplot()
    * scatterplot() é simples
    * o relplot() permite múltiplos gráficos

````python
tips = pd.DataFrame({
    "total_bill": [10.34, 21.01, 23.68, 14.83, 8.77, 26.88, 15.42, 18.90, 11.25, 30.14, 9.50, 22.70],
    "tip":        [1.66,  3.50,  3.31,  3.23,  2.00,  4.10,  2.80,  3.10,  1.90,  5.00,  1.50,  3.80],
    "size":       [2, 3, 2, 4, 2, 4, 3, 2, 1, 5, 2, 3],
    "smoker":     ["No","No","Yes","Yes","No","Yes","No","Yes","No","Yes","No","Yes"],
    "time":       ["Dinner","Dinner","Lunch","Dinner","Lunch","Dinner","Lunch","Dinner","Lunch","Dinner","Lunch","Dinner"],
})
tips
````

````python
sns.relplot(x="total_bill",y="tip",data=tips,kind="scatter")
plt.title("Gorjeta vs Fatura")
plt.xlabel("Gorjeta")
plt.ylabel("Fatura total")
plt.tight_layout()
plt.show()
````

<details>
  <summary>

  </summary>

- Define se os subgráficos aparecem lado a lado para comparar o eixo y, ou empilhados para comparar o eixo x o uso conjunto cria uma matriz que cruza duas variáveis.

</details>

## Gráficos Categóricos

<details>
    <summary>
        <font size='3',color='white'>
            <b>Tipos categóricos</b>
        </font>
    </summary>

- Tipos:
    * countplot
        + Contagem de observações por categoria.
    * barplot
        + Tendência central (média, mediana, etc.) de uma variável numérica por categoria. 
    * boxplot
        + É ideal para visualizar a distribuição de dados quantitativos, exibindo quartis, mediana e outliers
    * pointplot
        + Exibe a média de uma variável quantitativa para cada categoria, acompanhada de intervalos de confiança
</details>

### Countplot

````python
sns.catplot(x="time", data=tips, kind="count")
````

### Barplot

````python
sns.catplot(x="time", y="total_bill", data=tips, kind="bar")
````

### Boxplot

````python
sns.catplot(x="time", y="total_bill", data=tips, kind="box")
````

#### Remover outliers:

````python
sns.catplot(x="time", y="total_bill", data=tips, kind="box", showfliers=False)
````

#### Alterar whiskers:

````python
sns.catplot(x="time", y="total_bill", data=tips, kind="box", whis=[0,100])
````

### Pointplot

````python
sns.catplot(x="tip", y="total_bill", data=tips, kind="point")
````

````python
from numpy import median

sns.catplot(x="smoker", y="total_bill", data=tips, kind="point", estimator=median)
````

#### Síntese:

````python
import pandas as pd

df = pd.DataFrame({
    "salario": [3200, 3500, 3800, 4000, 3100,   # TI
                2200, 2400, 2600, 2500, 2300,   # RH
                5000, 5500, 5200, 4800, 5300,   # Financeiro
                1800, 2000, 1900, 2100, 1950],  # Operacional
    "setor":   ["TI"] * 5 + ["RH"] * 5 + ["Financeiro"] * 5 + ["Operacional"] * 5
})

ordem = df.groupby("setor")["salario"].median().sort_values().index.tolist()

sns.boxplot(x="setor", y="salario", data=df, order=ordem, showfliers=False)
````

## Estilo e Customização

<details>
    <summary>
        <font size='3',color='white'>
            <b>Tipos categóricos</b>
        </font>
    </summary>

- Estilo
    * O que controla?
        + Cor de fundo
        + Grade (grid)
        + aparência dos eixos
    * Como usar:
    ```python
            sns.set_style("whitegrid")
    ```
    * Opções:
        + white: fundo branco simples
        + dark: fundo escuro
        + ticks: estilo minimalista
        + whitegrid: fundo branco + grade
        + darkgrid: fundo escuro + grade
- Paleta
    * O que controla?
        + Cores principais do gráfico
    * Uso básico:
        ```python
            sns.set_palette("RdBu")
        ```
    * Tipos de paletas
        1. Qualitativa
            + Para categorias
            Ex: cores diferentes
        2. Sequencial
            + Para valores crescentes
            Ex: intensidade
        3. Divergente
            + Mostra extremos 
            Ex: positivo vs negativo
    * Paleta customizada:
    ```python
            sns.set_palette(["red", "blue", "green"])
    ```
- Context
    * O que muda?
        + Tamanho de:
        + fonte
        + linhas
        + elementos
    * paper: artigos
    * notebook: padrão
    * talk: apresentação
    * poster: muito grande
- Títulos
    * FacetGrid:
        + usado por:
            1. relplot()
            2. catplot()
            ```python
            grafico.figure.suptitle("Título")
            ```
    * Gráfico simples:
        + usado por:
            1. scatterplot()
            2. countplot()
    * Ajuste de posição
    ```python
        g.figure.suptitle("Título", y=1.03)
    ```
- Labels
    * Rótulos melhoram a interpretação.
    ```python
        g.set(xlabel="Eixo X",ylabel="Eixo Y")
    ```
- Rotação
    * Quando usar?
        1. nomes longos
        2. muitas categorias
    ```python
        plt.xticks(rotation=90)
    ```
</details>

#### Paleta

````python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset("tips")

sns.set_palette("Set2")

sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="day",
    data=tips
)

plt.title("Paleta padrão (Set2)")
plt.show()
````

````python
sns.set_palette(["red", "blue", "green", "purple"])

sns.scatterplot(
    x="total_bill",
    y="tip",
    hue="day",
    data=tips
)

plt.title("Paleta customizada")
plt.show()
````

#### Context

````python
sns.set_context("poster")
sns.scatterplot(x=[1,2,3], y=[4,5,6])
plt.show()
````

#### Titulos

````python
g = sns.catplot(
    x="day",
    y="total_bill",
    data=tips,
    kind="box"
)

g.figure.suptitle("Distribuição da Conta por Dia")

plt.show()
````

````python
g = sns.scatterplot(
    x="total_bill",
    y="tip",
    data=tips
)

g.set_title("Relação entre Conta e Gorjeta")

plt.show()
````

````python
g = sns.catplot(
    x="day",
    y="total_bill",
    data=tips,
    col="smoker",  # cria subplots
    kind="box"
)

g.set_titles("Fumante: {col_name}")

plt.show()
````

#### Rótulos

````python
g = sns.scatterplot(
    x="total_bill",
    y="tip",
    data=tips
)

g.set(
    xlabel="Conta Total (R$)",
    ylabel="Gorjeta (R$)"
)

plt.show()
````

#### Rotação

````python
g = sns.catplot(
    x="day",
    y="total_bill",
    data=tips,
    kind="bar"
)

plt.xticks(rotation=45)

plt.show()
````

## Juntando tudo

````python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "tempo":     ["Jan","Fev","Mar","Abr","Mai","Jun"] * 6,
    "vendas":    [150,180,170,210,230,260,
                  120,140,160,155,175,190,
                  200,220,210,240,260,280,
                  170,190,180,200,220,240,
                  130,150,140,160,180,200,
                  110,130,120,140,160,180],
    "categoria": (["Produto A"] * 6 + ["Produto B"] * 6) * 3,
    "grupo":     ["Grupo 1"] * 12 + ["Grupo 2"] * 12 + ["Grupo 3"] * 12,
})

g = sns.relplot(
    data=df,
    x="tempo", y="vendas",
    hue="categoria",
    col="grupo",
    kind="line",
    height=4, aspect=0.9,
)

g.figure.suptitle("Vendas por Categoria e Grupo ao Longo do Tempo", y=1.03, fontsize=13, fontweight="bold")
g.set_axis_labels("Mês", "Vendas (R$)")

for ax in g.axes.flat:
    ax.tick_params(axis="x", rotation=45)

plt.tight_layout()
````
