

```python
          df[df['country'].isin(['Italy','Portugal'])].head()
```
- Seleciona tipos de dados especificos e mostra em uma tabela.
```python
          df.select_dtypes("number").head()
```

<details>
  <summary>
    <font size='3', color='white'>
      <b>Agregar dados não agrupados .agg( )</b>
    </font>
  </summary>
 
- Agrega os dados em todas as linhas de uma determinada coluna.
```python
            df[['price','points']].agg(['mean','std'])
```
- Também é possivel usar um dicionário para passar informações.
```python
            df.agg({'price': ['std','mean'], 'points': ['mean']})
```
</details>

````python
df[['price','points']].agg(['mean','std'])
````

````python
df.agg({'price': ['std','mean'], 'points': ['mean']})
````

<details>
  <summary>
    <font size='3', color='white'>
      <b>União de .groupby() .agg( )</b>
    </font>
  </summary>
 
- É possível criar colunas com as agregações desejadas usando:
```python
            df.groupby("country").agg(
    mean_prices = ("price", "mean"),
    std_prices=("price", "std"),
    median_points=("points","mean")
)
```
</details>

````python
df.groupby("country").agg(
    mean_prices = ("price", "mean"),
    std_prices=("price", "std"),
    median_points=("points","mean")
).head()
````

````python
sns.barplot(data=df,x="points", y="price")
plt.show()
````
