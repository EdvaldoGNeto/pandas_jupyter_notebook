```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

imoveis = pd.read_csv('dados2.csv', sep=';')
imoveis
```

```python
gap = [50,150,200,250,300,350,450]

freq_bairro = imoveis['Bairro'].value_counts().reset_index()
freq_bairro.columns = ['Bairro','Frequencia']
freq_tipo = imoveis['Tipo'].value_counts().reset_index()
freq_tipo.columns = ['Tipo','Frequencia']
freq_quartos = imoveis['N de quartos'].value_counts().sort_index().reset_index()
freq_quartos.columns = ['N de quartos','Frequencia']
freq_preco = pd.cut(imoveis['Preço (mil)'], bins=gap).value_counts().sort_index().reset_index()
freq_preco.columns = ['Preço (mil)','Frequencia']
```

```python
df_bairro = pd.DataFrame(freq_bairro)
df_tipo = pd.DataFrame(freq_tipo)
df_quartos = pd.DataFrame(freq_quartos)
df_preco = pd.DataFrame(freq_preco)
```

```python
sns.barplot(data=df_bairro,x='Frequencia',y='Bairro',hue='Bairro')

plt.title('Frequência por bairro')
plt.show()
```

```python
sns.barplot(data=df_tipo,x='Frequencia',y='Tipo',hue='Tipo')

plt.title('Frequência por tipo')
plt.show()
```

```python
sns.barplot(data=df_quartos,x='N de quartos',y='Frequencia',hue='Frequencia',legend=False)

plt.title('Frequência por N de quartos')
plt.show()
```

```python
sns.barplot(data=df_preco,x='Preço (mil)',y='Frequencia',hue='Preço (mil)',legend=False)

plt.title('Frequência por preço')
plt.show()
```
