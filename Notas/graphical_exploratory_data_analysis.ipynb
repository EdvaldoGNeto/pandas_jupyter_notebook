````python
import pandas as pd

eleicoes = pd.read_csv("Presidente_BR_1T_2022.csv",encoding="latin_1",sep=";")
eleicoes_filtradas = eleicoes[['QT_VOTOS_TOTAL_ACUMULADO','LULA_QT_VOTOS_TOT','LULA_QT_VOTOS_TOT_ACUMULADO',
                               'JAIR_BOLSONARO_QT_VOTOS_TOT','JAIR_BOLSONARO_QT_VOTOS_TOT_ACUMULADO']]
eleicoes_filtradas['PORCENTAGEM_LULA'] = (eleicoes_filtradas['LULA_QT_VOTOS_TOT_ACUMULADO'] / eleicoes_filtradas['QT_VOTOS_TOTAL_ACUMULADO']) * 100
eleicoes_filtradas['PORCENTAGEM_JAIR_BOLSONARO'] = (eleicoes_filtradas['JAIR_BOLSONARO_QT_VOTOS_TOT_ACUMULADO'] / eleicoes_filtradas['QT_VOTOS_TOTAL_ACUMULADO']) * 100
eleicoes_filtradas
````

### Swarmplot

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# criando dados aleatórios
dados = {
    "PERCENTUAL_DESCONTO": np.random.randint(0, 101, 100),   # valores de 0 a 100%
    "QUANTIDADE_VENDIDA": np.random.randint(1, 500, 100)     # vendas entre 1 e 500
}

df = pd.DataFrame(dados)

df["FAIXA_DESCONTO"] = pd.cut(df["PERCENTUAL_DESCONTO"],bins=[0,20,40,60,80,100])

df.head()
````

````python
sns.swarmplot(data=df,x='FAIXA_DESCONTO',y='QUANTIDADE_VENDIDA')

plt.xlabel('Porcentagem de desconto (%)')
plt.ylabel('Quantidade vendida')
plt.show()
````

### ECDF (Empirical Cumulative Distribution Function / Função de Distribuição Acumulada Empírica)

<details>
    <summary>
        <font size='3',color='white'>
            <b>O que a ECDF responde?</b>
        </font>
    </summary>

- Ela mostra:
    * "Qual proporção dos dados é menor ou igual a determinado valor?"
- Como interpretar seu gráfico real?
    * Se no gráfico aparecer em 40% no eixo x e 0.3 no eixo y significa que: 
    **Cerca de 1.671 municípios (30% do Brasil) tiveram até 40% dos votos para Lula.**
    * Se aparecer em 70% no eixo x e 0.85 no eixo y significa que: 
    **Aproximadamente 4.734 municípios tiveram até 70% dos votos para Lula.**
</details>

### Qual porcentagem dos municípios teve até X% dos votos para Lula?

````python
x = np.sort(eleicoes_filtradas['PORCENTAGEM_LULA'])

n = 5.569
y = np.arange(1, len(x)+1) / n

_ = plt.plot(x, y, marker='.', linestyle='none')

_ = plt.xlabel('Porcentagem de votos para Lula')
_ = plt.ylabel('Proporção dos 5.569 municípios')
plt.margins(0.02) 
plt.show()
````

### Comparando dois ECDFs

````python
x_lula = np.sort(eleicoes_filtradas['PORCENTAGEM_LULA'])
y_lula = np.arange(1, len(x_lula)+1) / 5569

x_bolso = np.sort(eleicoes_filtradas['PORCENTAGEM_JAIR_BOLSONARO'])
y_bolso = np.arange(1, len(x_bolso)+1) / 5569

plt.plot(x_lula, y_lula, marker='.', linestyle='none', label='Lula')
plt.plot(x_bolso, y_bolso, marker='.', linestyle='none', label='Bolsonaro')

plt.xlabel('Percentual de votos')
plt.ylabel('Proporção de municípios')
plt.legend()
plt.show()
````

**No gráfico vemos quem teve distribuição mais concentrada entre municípios.**
