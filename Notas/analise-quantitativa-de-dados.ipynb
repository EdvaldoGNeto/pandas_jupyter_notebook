    <summary>
        <font style="font-size: 22px;color: #ffffff">
            <b>Média, Mediana e Moda</b>
        </font>
    </summary>
<font style='color: #ffffff'>

1. Média
    - A média soma todos os valores e divide pela quantidade.
        ```python
            np.mean()
        ```
        + calcula a média;
        + soma todos os valores e divide pela quantidade.
2. Mediana
    - A mediana é o valor central dos dados ordenados.
        ```python
            np.median()
        ```
        + calcula a mediana.
        + encontra o valor que fica exatamente no meio.
3. Moda
    - Moda é o valor mais frequente.
        ```python
            statistics.mode()
        ```
        + retorna o valor que mais aparece;
        + funciona com dados categóricos.

</font>
</details>

<details style="color: #b5f877">
    <summary>
        <font style="font-size: 22px;color: #ffffff">
            <b>Variância</b>
        </font>
    </summary>

<font style='color: #ffffff'>

- A variância mede o espalhamento dos dados em relação à média.
    
    ```python
    np.var(dataframe['coluna'], ddof=1)
    ```
    * **np.var()**
        + calcula a variância.
        + mede o quanto os dados variam.
    * **ddof=1**
        + usa a fórmula amostral.
        + divide por n-1 ao invés de n.

</font>
</details>

<details style="color: #b5f877">
    <summary>
        <font style="font-size: 22px;color: #ffffff">
            <b>Desvio padrão</b>
        </font>
    </summary>

<font style='color: #ffffff'>

- É a raiz quadrada da variância.
    
    ```python
    np.std(dataframe['coluna'], ddof=1)
    ```
    * **np.std()**
        + calcula o desvio padrão.
        + mede a dispersão dos dados.
    * **ddof=1**
        + usa o cálculo amostral.
        + deixa o resultado estatisticamente correto para amostras.
        
</font>
</details>

<details style="color: #b5f877">
    <summary>
        <font style="font-size: 22px;color: #ffffff">
            <b>Quantis e Quartis</b>
        </font>
    </summary>

<font style='color: #ffffff'>

- Quantis dividem os dados em partes.
    
    ```python
    np.quantile(dataframe['coluna'], [0, 0.25, 0.5, 0.75, 1])
    ```
    * **np.quantile()**
        + calcula quantis.
        + encontra posições específicas dos dados.
    * **[0, 0.25, 0.5, 0.75, 1]**
        + mínimo.
        + primeiro quartil.
        + mediana.
        + terceiro quartil.
        + máximo.
        
</font>
</details>
