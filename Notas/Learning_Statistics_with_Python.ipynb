<details>
  <summary>
    <font size='5', color='green'>
      <b>7.2.1 Convertendo contagens em proporções</b>
    </font>
  </summary>
 
- Salvar e renomear totais
```python
            tabs = pd.crosstab(index=df["speaker"], columns=df["utterance"],margins=True)

            tabs.columns = list(tabs.columns)[0:-1] + ['rowtotals'] #"All" vira 'rowtotals' por coluna
            tabs.index = list(tabs.index)[0:-1] + ['coltotals'] #"All" vira 'coltotals' por linha
```
- Proporção por coluna
```python
            tabs / tabs.loc['coltotals']
            # vai dividir todo tabs por coltotals
```
- Proporção por linha
```python
            tabs.div(tabs["rowtotals"], axis=0)
            # vai dividir todo tabs por rowtotals
```
</details>

````python
tabs = pd.crosstab(index=df["speaker"], columns=df["utterance"],margins=True)

tabs.columns = list(tabs.columns)[0:-1] + ['rowtotals']
tabs.index = list(tabs.index)[0:-1] + ['coltotals']

tabs
````

````python
tabs / tabs.loc['coltotals']
````

````python
tabs.div(tabs["rowtotals"], axis=0)
````

- As vezes é preciso transformar e reclassificar os dados, pois 
nem sempre estão no formato ideal.
- 7.3.1 Criando variáveis transformadas
  * Pegar um dado original e aplicar uma função matemática nele para criar uma nova versão mais útil.
  * O sistema fuciona como se fosse uma questão de opinião forte,neutra ou fraca.
  ```python
      data = [1, 7, 3, 4, 4, 4, 2, 6, 5, 5]
      # 1 - Opinião Fraca
      # 4 - Opinião Neutra
      # 7 - Opinião Forte
      data = [x-4 for x in data]
      # Após o código fica mais obvio
      [-3,  3, -1,  0,  0,  0, -2,  2,  1,  1]
      # Negativos: Fraco
      # 0 Neutro
      # Positivos: Forte
      
      #Usando
      data = np.sign(data)
      [-1,  1, -1,  0,  0,  0, -1,  1,  1,  1]
      # Fica ainda melhor de observar
      # -1 Fraco
      # 0 Neutro
      # 1 Forte
  ```
  * Força de Opinião
  ```python
      df['centered'] = df['scores']-4 # Transforma em opinião
      df['opinion_strength'] = abs(df['centered']) # O Nível de força da opinião
      df['opinion_direction'] = np.sign(df['scores']-4) # Exemplifica opinião positiva, negativa e neutra.
  ```
</details>

````python
data = [1, 7, 3, 4, 4, 4, 2, 6, 5, 5]
data = [x-4 for x in data]
data
````

````python
import numpy as np
data = np.array([1, 7, 3, 4, 4, 4, 2, 6, 5, 5])
data = data - 4
data
````

````python
data = np.array([1, 7, 3, 4, 4, 4, 2, 6, 5, 5])
data = data - 4
data = np.sign(data)
data
````

````python
df = pd.DataFrame(
    {'scores': np.array([1, 7, 3, 4, 4, 4, 2, 6, 5, 5])
    })

df
````

````python
df['centered'] = df['scores']-4
df['opinion_strength'] = abs(df['centered'])
df['opinion_direction'] = np.sign(df['scores']-4)
df
````

    <font size='5', color='green'>
      <b>7.3.2 Transformar números em categorias</b>
    </font>
  </summary>

- Usando cut você pode definir intervalos e transformar em categorias. 
```python
            df['categories'] = pd.cut(x = df['age'], bins = [0,20,40,60], labels = ['young', 'adult', 'older'])
            # O objetivo é transformar idade em grupos.
            # 0–20 -> jovem
            # 21–40 -> adulto
            # 41–60 -> mais velho
```
- Cortando em intervalos iguais ele é automatico.
```python
            pd.cut(df['age'], bins=3)
            # divide em intervalos iguais em tamanho
```
- Quantidade de elementos equilibradas
```python
            pd.qcut(df['age'], q=[0, .33, .66, 1])
            # Divide em grupos com mesma quantidade de dados percentuais
```

</details>

````python
age = [60,58,24,26,34,42,31,30,33,2,9]
df = pd.DataFrame(
    {'age': np.array([60,58,24,26,34,42,31,30,33,2,9])
    })

df
````

````python
df['categories'] = pd.cut(x = df['age'], bins = [0,20,40,60], labels = ['young', 'adult', 'older'])
df
````

````python
pd.cut(df['age'], bins=3)
````

````python
pd.qcut(df['age'], q=[0, .33, .66, 1])
````
