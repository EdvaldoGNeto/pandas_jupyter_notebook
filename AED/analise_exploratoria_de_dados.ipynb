### Apagando pasta criada e movendo o .csv

```python
import shutil
import os

pastas = [
    "infracoes_2018", "infracoes_2021", "infracoes_2025"
]

for pasta in pastas:
  pasta_raiz = f"/content/infracoes/{pasta}"
  # Verifica se há uma subpasta na pasta atual
  subdir = [name for name in os.listdir(pasta_raiz) if os.path.isdir(os.path.join(pasta_raiz, name))]
  if subdir:
    subpasta = f"{pasta_raiz}/{subdir[0]}"
    for arquivo in os.listdir(subpasta):
      if arquivo.endswith(".csv"):
        origem = os.path.join(subpasta, arquivo)
        destino = os.path.join(pasta_raiz, arquivo)
        shutil.move(origem, destino)
    # Remove a subpasta
    shutil.rmtree(subpasta)
```

## Infrações

### Juntando meses de  infrações.

```python
import pandas as pd
import shutil
import os

# Localizando pasta
lista_infracoes = []
pastas = [
    "infracoes_2018", "infracoes_2021", "infracoes_2025"
]


for dir in pastas:
  for raiz, pasta, arquivo in os.walk(f"/content/infracoes/{dir}"):
    if arquivo:
      lista_infracoes.append(sorted(arquivo))

aux = 0
lista_infracoes_2018 = []
lista_infracoes_2021 = []
lista_infracoes_2025 = []

# Colocando em listas

for infracao in lista_infracoes:
  for ordem, arquivo in enumerate(infracao):
    if aux<12:
      temp = pd.read_csv(f"/content/infracoes/infracoes_2018/{arquivo}",encoding="latin_1", sep=",")
      lista_infracoes_2018.append(temp)
      print(f"\033[34mDataframe {ordem+1} de 2018 lido!\033[m")
    elif 12<=aux<24:
      temp = pd.read_csv(f"/content/infracoes/infracoes_2021/{arquivo}",encoding="latin_1", sep=";")
      lista_infracoes_2021.append(temp)
      print(f"\033[34mDataframe {ordem+1} de 2021 lido!\033[m")
    elif 24<=aux:
      temp = pd.read_csv(f"/content/infracoes/infracoes_2025/{arquivo}",encoding="latin_1", sep=";")
      lista_infracoes_2025.append(temp)
      print(f"\033[34mDataframe {ordem+1} de 2025 lido!\033[m")
    aux +=1




lista_infracoes_2018 = pd.concat(lista_infracoes_2018)
lista_infracoes_2021 = pd.concat(lista_infracoes_2021)
lista_infracoes_2025 = pd.concat(lista_infracoes_2025)
```
