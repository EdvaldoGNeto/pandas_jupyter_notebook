    - Tipo: Qualitativa Nominal

2. <b>fase_dia</b>
    - Tipo: Qualitativa Ordinal
    - Existe uma ordem baseada no momento do dia (Pleno dia, Anoitecer, Plena noite).
    - Também aparece como string.
3. <b>idade</b>
    - Tipo: Quantitativa Discreta
    - Idade normalmente é contada em inteiros. Logo é discreta.
    - são classificados como int
4. <b>tipo_veiculo</b>
    - Tipo: Qualitativa Nominal
    - São descritos os nomes dos veiculos.
    - Também aparece como string.

```python
import pandas as pd

# Criando um mapeamento de cada string para se tornar int

mapping = {
    "Plena Noite": 0,
    "Anoitecer": 1,
    "Amanhecer": 2,
    "Pleno dia": 3
}

# Criando uma coluna codificada

df["fase_dia_encoded"] = df["fase_dia"].map(mapping)
df["fase_dia_encoded"].head()
```

```python
# Verificando se existe alguma string não mapeada

df[df["fase_dia"].map(mapping).isna()]["fase_dia"].unique()
```

<details>
    <summary>
        <font style='font-size:20px;color: #ffffff'>
            <b>Binning (faixa_etaria)</b>
        </font> 
    </summary>

- Transforma variável numérica em qualitativa ordinal
- Ajuda a identificar perfis demográficos
</details>

```python
import pandas as pd

bins = [0, 18, 30, 60, 120]
labels = ["Menores de idade", "Jovens", "Adultos", "Idosos"]

df["faixa_etaria"] = pd.cut(df["idade"], bins=bins, labels=labels, right=False)
df["faixa_etaria"]
```

<div>
    <b>
        <font style='font-size: 24px;color: #87ecfd'>
            Criando um novo dataframe com base na coluna "faixa_etaria".
        </font>
    </b>
</div>

```python
import pandas as pd

df_freq = df["faixa_etaria"].value_counts().reset_index()
df_freq.columns = ["faixa_etaria", "fi"]
df_freq
```

<div>
    <b>
        <font style='font-size: 24px;color: #87ecfd'>
            Ordenando as categórias do dataframe
        </font>
    </b>
</div>

```python
ordem = ["Menores de idade", "Jovens", "Adultos", "Idosos"]

df_freq["faixa_etaria"] = pd.Categorical(df_freq["faixa_etaria"], categories=ordem, ordered=True)
df_freq = df_freq.sort_values("faixa_etaria").reset_index(drop=True)
df_freq
```

<div>
    <b>
        <font style='font-size: 24px;color: #87ecfd'>
            Definindo a coluna acumulativa do dataframe.
        </font>
    </b>
</div>

```python
df_freq["Fi"] = df_freq["fi"].cumsum()
df_freq
```

<div>
    <b>
        <font style='font-size: 24px;color: #87ecfd'>
            Definindo as colunas de porcentagem do dataframe.
        </font>
    </b>
</div>

```python
total = df_freq["fi"].sum()

df_freq["fr (%)"] = (df_freq["fi"]/total) *100
df_freq["Fr (%)"] = df_freq["fr (%)"].cumsum()

df_freq
```

<div>
    <b>
        <font style='font-size: 24px;color: #87ecfd'>
            Arredondando o números quebrados da divisão. Além disso transformando em string para poder adicionar o "%"
        </font>
    </b>
</div>

```python
df_freq["fr (%)"] = df_freq["fr (%)"].round(0).astype(str) + "%"
df_freq["Fr (%)"] = df_freq["Fr (%)"].round(0).astype(str) + "%"

df_freq
```

```python
linha_nova = {"faixa_etaria": "total","fi": total,"fr (%)":"100%"}
df_freq.loc[len(df_freq)] = linha_nova
df_freq
```
