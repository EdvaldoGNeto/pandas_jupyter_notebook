````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

**1. Frequências**
</div>

<a id="freq-abs"></a>
<div style="font-size:30px">

**1.1 Frequência Absoluta**
</div>

````python
gap = [0, 1000, 10000, 50000, 500000]

freq_abs = pd.cut(df2025["Incêndio (ha)"],bins=gap)
freq_abs = freq_abs.value_counts().sort_index()
freq_abs
````

````python
sns.histplot(data=df2025["Incêndio (ha)"],bins=5,binrange=[0,500000])

plt.xlabel("Área atingida (ha)")
plt.ylabel("Frequência Absoluta")
plt.title("Área atingida por frequência absoluta")

plt.show()
````

<a id="freq-rel"></a>
<div style="font-size:30px">

**1.2 Frequência Relativa**
</div>

````python
freq_rel = round((freq_abs/freq_abs.sum()) * 100)
freq_rel
````

````python
sns.histplot(data=df2025,x="Incêndio (ha)",stat="percent")

plt.xlabel("Área atingida (ha)")
plt.ylabel("Frequência Relativa")
plt.title("Área atingida por frequência relativa")

plt.show()
````

<div style="font-size:30px">

**1.2.1 Comparando Frequência Absoluta e Relativa**
</div>

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df2025["Incêndio (ha)"], bins=5, binrange=[0, 500000], ax=axes[0])
axes[0].set_title("Contagem absoluta")

sns.histplot(data=df2025, x="Incêndio (ha)", stat="percent", ax=axes[1])
axes[1].set_title("Contagem relativa")

plt.tight_layout()
plt.show()
````

<a id="freq-acum"></a>
<div style="font-size:30px">

**1.3 Frequência Acumulada**
</div>

````python
freq_acum = freq_abs.cumsum()
freq_acum
````

````python
sns.histplot(data=df2025,x=df2025["Incêndio (ha)"],stat="percent",cumulative=True)
````

````python
freq_acum.plot()
````

<a id="freq-acum"></a>
<div style="font-size:30px">

**1.3.1 Observando qual o melhor gráfico para frequência acumulada**
</div>

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df2025,x=df2025["Incêndio (ha)"],stat="percent",cumulative=True,ax=axes[0])
axes[0].set_title("Frequência acumulada em Hist")

freq_acum.plot(ax=axes[1])
axes[1].set_title("Frequência acumulada em Line")

plt.tight_layout()
plt.show()
````

**2. Média,Moda e Mediana**
</div>

````python
media = df2025["Incêndio (ha)"].mean()
mediana = df2025["Incêndio (ha)"].median()
moda = df2025["Incêndio (ha)"].mode()

print(f"Média: {media:.2f}\n"
      f"Mediana: {mediana:.2f}\n"
      f"Moda: {moda}\n")
````

<div>
<font style="font-size:20px">

**Os resultados indicam diferenças entre média, mediana e moda, sugerindo que os dados não apresentam distribuição simétrica. A média tende a assumir valores mais elevados devido à presença de incêndios muito extensos, caracterizando influência de valores extremos.**

**A mediana representa de forma mais adequada o comportamento geral dos dados, pois é menos sensível a valores extremos e reflete melhor a concentração da maior parte das observações em áreas menores.**

**Já a moda corresponde ao valor mais frequente da distribuição, indicando predominância de incêndios de pequena extensão.** 

**Portanto, a distribuição apresenta assimetria positiva, com concentração de valores baixos e poucos valores muito altos.**


</font>
</div>

<a id="dispersao"></a>
<div class="alert alert-success"style="font-size:40px">

**3. Variância, Desvio padrão e Interquartil**
</div>

````python
variancia = df2025["Incêndio (ha)"].var()
desvio = df2025["Incêndio (ha)"].std()

print(f"Variância: {variancia:.2f}\n"
      f"Desvio padrão: {desvio:.2f}")
````

````python
Q1 = df2025["Incêndio (ha)"].quantile(0.25)
Q3 = df2025["Incêndio (ha)"].quantile(0.75)

iqr = Q3 - Q1

print(f"Interquartil: {iqr:.2f}")
````

<div>
<font style="font-size:20px">

**As medidas de dispersão indicam elevada variabilidade nos dados referentes à área atingida por incêndios. A variância e o desvio padrão apresentaram valores elevados, evidenciando grande dispersão em relação à média.**

**O intervalo interquartil mostra a amplitude da metade central dos dados, indicando que a maior parte das observações está concentrada em valores menores, enquanto poucos registros apresentam áreas extremamente elevadas.**

**Esses resultados sugerem que o conjunto de dados é heterogêneo, pois existe grande diferença entre os valores observados. A presença de incêndios de grande extensão contribui significativamente para o aumento da dispersão e reforça a assimetria positiva da distribuição.**



</font>
</div>

**4. Comparação Estatística**
</div>

````python
variaveis = [
    "Queima prescrita  (ha) ",
    "Queima controlada  (ha) ",
    "Gestão de ignição natural (ha) ",
    "Gestão de ignição antrópica  (ha) "
]

for col in variaveis:
    print("=========================================================")
    print(col,"\n")

    print("Média:", df2025[col].mean())
    print("Mediana:", df2025[col].median())
    print("Moda:")
    print(df2025[col].mode())

    print("Variância:", df2025[col].var())
    print("Desvio padrão:", df2025[col].std())

    Q1 = df2025[col].quantile(0.25)
    Q3 = df2025[col].quantile(0.75)

    print("IQR:", Q3 - Q1)
````

````python
sns.boxplot(data=df2025[variaveis])
plt.xticks(rotation=15)
plt.show()
````

<div>
<font style="font-size:20px">

**O gráfico não permite uma análise visual adequada então baseado nas variáveis analisadas. Algumas estratégias apresentam médias e medianas mais elevadas, indicando maior área manejada de forma recorrente. As medidas de dispersão evidenciam elevada variabilidade em determinadas estratégias, principalmente devido à presença de valores extremos. As variáveis com maiores desvios padrão e intervalos interquartis podem ser consideradas as mais heterogêneas, indicando forte variação entre os registros observados.**



</font>
</div>

**6. Comparação de Boxplot**
</div>

````python
df_2022 = pd.read_excel("planilha_de_area_atingida_por_fogo_historico_dgeo_2025.xlsx",sheet_name="2022")
df_2023 = pd.read_excel("planilha_de_area_atingida_por_fogo_historico_dgeo_2025.xlsx",sheet_name="2023")
df_2024 = pd.read_excel("planilha_de_area_atingida_por_fogo_historico_dgeo_2025.xlsx",sheet_name="2024")
````

````python
dfs = [df_2022,df_2023,df_2024]
anos = [2022,2023,2024]

for ordem,df in enumerate(dfs):
    print(f"================{anos[ordem]}================")
    df.info()
````

````python
df_2022 = df_2022.rename(columns={'Incêndio (ha) ': 'Incêndio (ha)'})   
````

````python
df_2022["Ano"] = "2022"
df_2023["Ano"] = "2023"
df_2024["Ano"] = "2024"

dados = pd.concat(
    [df_2022, df_2023, df_2024],
    ignore_index=True
)

plt.figure(figsize=(12,8))

sns.boxplot(
    data=dados,
    x="Ano",
    y="Incêndio (ha)"
)

plt.yscale("log")

plt.title("Comparação dos incêndios entre 2022, 2023 e 2024")

plt.show()
````

<div>
<font style="font-size:20px">

**Os boxplots comparativos entre os anos evidenciam diferenças relevantes nas distribuições das áreas queimadas. Alguns anos apresentam medianas mais elevadas, indicando aumento típico das áreas afetadas por incêndios, determinados anos mostram maior amplitude e presença mais intensa de valores extremos, sugerindo ocorrência de incêndios de grande magnitude. Os anos com caixas mais dispersas e maior quantidade de outliers podem ser considerados os mais críticos em relação às áreas queimadas.**



</font>
</div>

    * Mostra as 5 primeiras linhas da tabela.
        + verificar se a leitura deu certo.
        + visualizar as colunas.
        + entender a estrutura dos dados.

    * Calcula estatísticas automáticas das colunas numéricas, usado para obter uma visão geral dos dados.
        + média.

        + mínimo.
        + máximo.
        + quartis.
- **df2025.info()**

<div style="font-size:30px">

**Frequência Absoluta**
</div>

- **gap**
    * Define os intervalos:
        + 0 até 1000
        + 1000 até 10000
        + 10000 até 50000
        + 50000 até 500000
- **pd.cut()**
    * Divide os dados em classes/faixas.
    * Exemplo:
        + um valor 7000 entra no intervalo 1000–10000.
- **value_counts()**
    * Conta quantos valores existem em cada faixa.
- **sort_index()**
    * Organiza os intervalos em ordem crescente.
    * Para que foi usado?
        + Para construir tabela de frequência absoluta. 
        + histograma.
- **sns.histplot()**

    * Cria histograma.
    * bins=5
        + Divide o gráfico em 5 faixas.
    * binrange=[0,500000]
        + Define o intervalo total do gráfico.
- **plt.xlabel()**
    * Nome do eixo X.
- **plt.ylabel()**
    * Nome do eixo Y.
- **plt.title()**
    * Título do gráfico.
- **plt.show()**
    * Exibe o gráfico.


<div style="font-size:30px">

**Frequência Relativa**
</div>

- **freq_abs.sum()**
    * Soma todas as frequências.
- **freq_abs / freq_abs.sum()**
    * Transforma em proporção.
- **Multiplica por 100**
    * Transforma em porcentagem.
- **round()**
    * Arredonda os valores.
- **stat="percent"**
    * Troca o eixo Y:
        + de contagem.
        + para porcentagem.
