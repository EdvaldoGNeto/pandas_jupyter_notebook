**Questão 2 — Média Amostral em Python**
</div>

<details style="color: #b5f877">
    <summary>
        <font style="font-size: 22px;color: #ffffff">
            <b>Média Amostral</b>
        </font>
    </summary>
<font style='color: #000000'>

```python
    dados = [12, 15, 18, 20, 22, 25]
```

- Média amostral manual:
$$x = \frac{12 + 15 + 18 + 20 + 22 + 25}{6} = \frac{112}{6} \approx 18.67$$

- Média usando códigos:
```python
    np.mean()
```

</font>

````python
dados = [12, 15, 18, 20, 22, 25]
media = np.mean(dados)

# Criando o gráfico
sns.lineplot(x=range(len(dados)), y=dados, marker="o",color="black")

# Linha da média
plt.axhline(media, linestyle="--", label=f"Média = {media:.2f}",color="red")

plt.xlabel("Índice")
plt.ylabel("Valores")

plt.grid()
plt.legend()
plt.show()
````

**Questão 4 — Simulação de Médias Amostrais**
</div>

````python
np.random.seed(28)

medias_amostrais = []

populacao = np.random.normal(loc=50,scale=10,size=1000)

for i in range(100):

    amostra = np.random.choice(populacao,30,replace=False)

    media = np.mean(amostra)

    medias_amostrais.append(media)

media_total = np.mean(medias_amostrais)

print(f"Média: {media_total:.2f}")
````

````python
sns.histplot(medias_amostrais,bins=10,kde=True,color="black")

plt.axvline(media_total, linestyle="--",label=f"Média = {media_total:.2f}",color="green")

plt.xlabel("Médias Amostrais")
plt.ylabel("Frequência")
plt.grid()
plt.legend()
plt.show()
````

<div class='alert alert-success' style='font-size:24px'>

**Questão 5 — Distribuição Amostral**
</div>

<div class='alert alert-info' style='font-size:20px'>

- **O que é:** A distribuição amostral da média é a distribuição de probabilidade de todas as médias possíveis calculadas a partir de amostras de mesmo tamanho extraídas de uma população. 

- **Importância para a Inferência:** Ela é a base de toda a inferência estatística. ela mostra como as médias das amostras se comportam e quão distantes elas tendem a estar do verdadeiro parâmetro populacional, permitindo quantificar a incerteza de nossas estimativas.
</div>

<div class='alert alert-success' style='font-size:24px'>

**Questão 6 — Intervalo de Confiança**
</div>

$$IC_{95\%} = [12.4, 18.7]$$

<div class='alert alert-info' style='font-size:20px'>

- **Significado:** Um intervalo de confiança de 95% indica que, se repetíssemos o processo de amostragem e cálculo do intervalo muitas vezes, aproximadamente 95 desses intervalos gerados conteriam o verdadeiro parâmetro populacional. 

- **Interpretação de $IC_{95\%}$:** Estimamos com 95% de confiança que o verdadeiro parâmetro populacional está compreendido entre os valores 12.4 e 18.7. Não significa que há 95% de chance de o parâmetro estar ali dentro, mas sim que o método de construção do intervalo funciona 95% das vezes.
</div>

**Questão 7 — Erro Padrão**
</div>

$$dados = [5, 8, 9, 10, 12, 15, 18]$$

````python
dados = [5, 8, 9, 10, 12, 15, 18]
desvio_padrao_amostral = np.std(dados)
n = len(dados)
erro_padrao = desvio_padrao_amostral / np.sqrt(n)

print(f"a) Desvio padrão amostral: {desvio_padrao_amostral:.2f}")
print(f"b) Erro padrão da média: {erro_padrao:.2f}")
````

<div class='alert alert-success' style='font-size:24px'>

**Questão 8 — Interpretação do Erro Padrão**
</div>

<div class='alert alert-info' style='font-size:20px'>

**O erro padrão diminui quanto maior a amostra, pois quanto maiores as amostras mais informação e isso reduz o erro, tornando a média mais estável e precisa.**

</div>

**Questão 9 — Teorema Central do Limite**
</div>

````python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(28)

tamanho_amostra = 40
num_amostras = 100

medias_exponencial = []

for i in range(num_amostras):
    amostra = np.random.exponential(scale=2, size=tamanho_amostra)
    media = np.mean(amostra)
    medias_exponencial.append(media)

plt.hist(medias_exponencial, bins=20, edgecolor='black', color='skyblue')

plt.title("Distribuição das Médias Amostrais Exponencial")
plt.xlabel("Médias Amostrais")
plt.ylabel("Frequência")

plt.grid(alpha=0.8)
plt.show()
````

<div class='alert alert-info' style='font-size:20px'>

**Mesmo que a distribuição original dos dados seja altamente assimétrica, o histograma das médias amostrais apresenta uma Distribuição Normal. Isso ilustra perfeitamente o Teorema Central do Limite: à medida que o tamanho amostral aumenta, a distribuição das médias das amostras tende a uma distribuição normal.**

</div>

**Questão 11 — Intervalo de Confiança com SciPy**
</div>

````python
dados = [22, 25, 21, 20, 26, 24, 23, 27]

media_amostral = np.mean(dados)
erro_padrao = st.sem(dados) 
graus_liberdade = len(dados) - 1

ic = st.t.interval(confidence=0.95, df=graus_liberdade, loc=media_amostral, scale=erro_padrao)

print(f"Média amostral: {media_amostral:.2f}")
print(f"Intervalo de Confiança de 95%: [{ic[0]:.2f}, {ic[1]:.2f}]")
````

- **Explicação:** A precisão de um intervalo de confiança é medida pela sua distância entre o limite superior e o inferior. Quanto menor a distância e o intervalo, menor é a margem de erro envolvida e mais precisa é a estimativa.

**Questão 13 — Variabilidade das Médias**
</div>

````python
np.random.seed(28)
populacao_base = np.random.normal(loc=50, scale=10, size=1000)

medias_n5 = []

for i in range(50):
    amostra = np.random.choice(populacao_base, size=5)
    media = np.mean(amostra)
    medias_n5.append(media)


medias_n100 = []

for i in range(50):
    amostra = np.random.choice(populacao_base, size=100)
    media = np.mean(amostra)
    medias_n100.append(media)

var_n5 = np.std(medias_n5)
var_n100 = np.std(medias_n100)

print(f"Desvio Padrão das médias (5): {var_n5:.2f}")
print(f"Desvio Padrão das médias (100): {var_n100:.2f}")
````

````python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

sns.histplot(medias_n5, bins=20, kde=True, ax=axes[0], color="blue")
axes[0].set_title("Distribuição das médias (5)")
axes[0].set_xlabel("Médias amostrais")
axes[0].set_ylabel("Frequência")
axes[0].grid(True,alpha=0.8)

sns.histplot(medias_n100, bins=20, kde=True, ax=axes[1], color="green")
axes[1].set_title("Distribuição das médias (100)")
axes[1].set_xlabel("Médias amostrais")
axes[1].set_ylabel("Frequência")
axes[1].grid(True,alpha=0.8)
plt.show()
````

<div class='alert alert-info' style='font-size:20px'>

- **a variabilidade das médias diminui drasticamente. As médias calculadas a partir de amostras de tamanho 100 ficam muito mais concentradas próximas à média real da população.**
- **Já as de tamanho 5 espalham-se muito mais, apresentando alta variabilidade.**

</div>

**Questão 14 — Estimativa Pontual e Intervalar**
</div>

<div class='alert alert-info' style='font-size:20px'>

- **Estimativa Pontual:** É um único valor numérico calculado a partir da amostra para estimar diretamente o parâmetro populacional.

    * *Exemplo:* Com base na mesma pesquisa, conclui-se com 95% de confiança que o tempo médio diário de estudo de todos os estudantes está entre 3,8 e 4,6 horas.

- **Estimativa Intervalar:** É uma faixa de valores dentro da qual se espera que o verdadeiro parâmetro populacional esteja contido, associada a uma probabilidade ou nível de confiança.

    * *Exemplo:* A partir das medições, estima-se com 90% de confiança que a temperatura média diária real da região está entre 27°C e 29°C.

</div>
